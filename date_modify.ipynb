```python
functions = [{
        'name': 'get_item_info',
        'description': '取得時間範圍內某個產品的資訊',
        'parameters': {
            'type': 'object',
            'properties': {
                'item': {
                    'type': 'string',
                    'description':
                    '產品名稱',
                },
                'start': {
                    'type': 'string',
                    'description':
                    '時間範圍的起始時間'
                },
                'end': {
                    'type': 'string',
                    'description':
                    '時間範圍的結束時間'
                },
            },
            'required': ['event', 'start', 'end'],
        }
    }]
```

```python
import jionlp
query = '去年中秋節的營收多少'
def date_extract(query):
    date = jionlp.parse_time(query)
    return date['time']
date = date_extract(query)
print(date)
```

```python
date_modify_prompt = '''
你是一個日期轉換助手，請按照如下格式對使用者輸入進行轉化。

使用者輸入：
2023年端午節的訂單銷量是多少

日期：
['2023-06-22 00:00:00', '2023-06-22 23:59:59']

輸出：
2023-06-22 00:00:00到2023-06-22 23:59:59的訂單銷量是多少

使用者輸入：
2022年中秋節的營收多少

日期：
['2022-09-10 00:00:00', '2022-09-10 23:59:59']

輸出：
2022-09-10 00:00:00到2022-09-10 23:59:59營收多少

使用者輸入：
{}

日期：
{}

輸出：
'''
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(temperature=0, model="qwen2", api_key="nn", base_url='http://10.250.2.23:8600/v1')
```

```python
query = '去年第一季度的銷量怎麼樣'
modify_query = llm.invoke(date_modify_prompt.format(query, date_extract(query))).content
print(modify_query)
```
